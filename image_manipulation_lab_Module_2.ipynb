**art 2: Cropping Images**

```python
# Load color image
img_Elijah_Ahmir = cv2.imread('Mod2/visuals/Ahmir_n_Eli.png', cv2.IMREAD_COLOR)

# Display it (note: OpenCV uses BGR, matplotlib uses RGB)
plt.figure(figsize=(8, 8))
plt.imshow(img_Elijah_Ahmir[:, :, ::-1])
plt.show()
```

```python
# Crop out the middle region
# Format: [rows (y-axis), columns (x-axis)]
cropped_region = img_Elijah_Ahmir[20:420, 150:600]

# Display the cropped image
plt.imshow(cropped_region[:, :, ::-1])
plt.show()
```

**Step 3.1: Method 1 - Resize Using Scale Factors**

```python
resized_2x = cv2.resize(cropped_region, None, fx=2, fy=2)

# Display the resized image
plt.imshow(resized_2x[:, :, ::-1])
plt.show()
```

**Step 3.2: Method 2 - Resize to Exact Dimensions**

```python
# Set exact dimensions
desired_width = 400
desired_height = 200
dim = (desired_width, desired_height)

# Resize to exact dimensions
resized_exact = cv2.resize(cropped_region, dsize=dim, interpolation=cv2.INTER_AREA)

plt.imshow(resized_exact[:, :, ::-1])
plt.show()
```

Step 3.3: Method 3 - Resize While Maintaining Aspect Ratio


```python
# Specify desired width
desired_width = 200

# Calculate aspect ratio
aspect_ratio = cropped_region.shape[1] / cropped_region.shape[0]

# Calculate height based on aspect ratio
desired_height = int(desired_width / aspect_ratio)

# Create dimension tuple
dim = (desired_width, desired_height)

# Resize with maintained aspect ratio
resized_aspect = cv2.resize(cropped_region, dsize=dim, interpolation=cv2.INTER_AREA)

plt.imshow(resized_aspect[:, :, ::-1])
plt.show()
```

Step 4.1: Apply All Three Flip Types

```python
# Horizontal flip (mirror image)
img_flipped_horz = cv2.flip(img_Elijah_Ahmir, 1)

# Vertical flip (upside down)
img_flipped_vert = cv2.flip(img_Elijah_Ahmir, 0)

# Both flips (180-degree rotation)
img_flipped_both = cv2.flip(img_Elijah_Ahmir, -1)

# Display all versions
plt.figure(figsize=(18, 5))
plt.subplot(141); plt.imshow(img_flipped_horz[:, :, ::-1])
plt.title('Horizontal Flip')
plt.subplot(142); plt.imshow(img_flipped_vert[:, :, ::-1])
plt.title('Vertical Flip')
plt.subplot(143); plt.imshow(img_flipped_both[:, :, ::-1])
plt.title('Both Flipped')
plt.subplot(144); plt.imshow(img_Elijah_Ahmir[:, :, ::-1])
plt.title('Original')
plt.show()
```

**Part 2: OpenCV Image Annotations**

```python
import cv2
import matplotlib.pyplot as plt

# Use the already loaded image variable `img_Elijah_Ahmir`
image = img_Elijah_Ahmir

# Display the original image
plt.figure(figsize=(7, 7))
# OpenCV loads images in BGR format, matplotlib expects RGB. Flip channels for correct display.
plt.imshow(image[:, :, ::-1])
plt.show()

# Check image shape (height, width, channels)
print(image.shape) # Should show 3 channels
```

**Step 1.3: Define Color Palette**

Define your colors as BGR tuples (Blue, Green, Red). Remember, OpenCV uses BGR order, not RGB!



```python
# Define colors (BGR format)
yellow = (0, 255, 255) # B=0, G=255, R=255
red = (0, 0, 255) # B=0, G=0, R=255
magenta = (255, 0, 255) # B=255, G=0, R=255
green = (0, 255, 0) # B=0, G=255, R=0
```

```python
# Make a copy to preserve original
image_line = image.copy()

# Draw a yellow line from (0, 445) to (450, 465)
image_line = cv2.line(image_line, (0, 445), (450, 465),
                      yellow, thickness=3, lineType=cv2.LINE_8)

# Display the annotated image
plt.figure(figsize=(10, 10))
plt.imshow(image_line[:, :, ::-1]) # Convert BGR to RGB for display
plt.show()
```

```python
# Make a fresh copy
image_line_aa = image.copy()

# Draw the same line with anti-aliasing
image_line_aa = cv2.line(image_line_aa, (0, 445), (450, 465),
                        yellow, thickness=3, lineType=cv2.LINE_AA)

plt.imshow(image_line_aa[:, :, ::-1])
plt.show()
```

```python
# Make a copy
image_circle = image.copy()

# Draw a red circle at (195, 55) with radius 20
image_circle = cv2.circle(image_circle, (195, 55), 20,
                          red, thickness=2, lineType=cv2.LINE_AA)

# Display result
plt.figure(figsize=(8, 8))
plt.imshow(image_circle[:, :, ::-1])
plt.show()
```
