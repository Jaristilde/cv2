**Step 3.2: Method 2 - Resize to Exact Dimensions**

Step 3.3: Method 3 - Resize While Maintaining Aspect Ratio


```python
# Specify desired width
desired_width = 200

# Calculate aspect ratio
aspect_ratio = cropped_region.shape[1] / cropped_region.shape[0]

# Calculate height based on aspect ratio
desired_height = int(desired_width / aspect_ratio)

# Create dimension tuple
dim = (desired_width, desired_height)

# Resize with maintained aspect ratio
resized_aspect = cv2.resize(cropped_region, dsize=dim, interpolation=cv2.INTER_AREA)

plt.imshow(resized_aspect[:, :, ::-1])
plt.show()
```
