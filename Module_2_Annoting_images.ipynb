```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
import requests

# Using a working URL for an Apollo 11 Launch image from Wikimedia Commons
image_url = 'https://upload.wikimedia.org/wikipedia/commons/thumb/d/d4/Apollo_11_launch_-_GPN-2000-000630.jpg/1280px-Apollo_11_launch_-_GPN-2000-000630.jpg'
image_filename = 'Apollo-11-Launch.jpg' # Renaming the file for clarity

response = requests.get(image_url)
if response.status_code == 200:
    with open(image_filename, 'wb') as f:
        f.write(response.content)
    print(f"Image '{image_filename}' downloaded successfully.")
else:
    print(f"Failed to download image. Status code: {response.status_code}")
```

```python
# Read image as color (3 channels)
image = cv2.imread('Apollo-8-Launch.jpg', cv2.IMREAD_COLOR)

# Check if image was loaded successfully
if image is None:
    print("Error: Could not load the image. The file 'Apollo-8-Launch.jpg' might not exist or is corrupted.")
else:
    # Display the original image (convert BGR to RGB for matplotlib)
    plt.figure(figsize=(7, 7))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.show()

    # Check image shape (height, width, channels)
    print(image.shape) # Should show 3 channels
```

```python
# Convert to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
plt.figure(figsize=(7, 7))
plt.imshow(gray_image, cmap='gray')
plt.show()

# Check image shape (height, width)
print(gray_image.shape) # Should show 2 channels
```

```python
# Apply a binary threshold
# Pixels above 127 become 255 (white), below 127 become 0 (black)
ret, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)

# Display the binary image
plt.figure(figsize=(7, 7))
plt.imshow(binary_image, cmap='gray')
plt.show()

# Check image shape (height, width)
print(binary_image.shape)
```
